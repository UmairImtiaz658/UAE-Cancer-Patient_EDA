```python
# check number of unique ages in age category
print('Count of unique ages :', df['Age'].nunique())

print('\n')
# check for unique ages in age category
print('Unique ages :', df['Age'].unique())
```

```python
# Create age categories before selecting
df['Age_Cat'] = pd.cut(df['Age'], bins=[0, 18, 40, 60, 100], 
                        labels=['Child', 'Adult', 'Senior', 'Elderly'])

# Now this will work:
df[['Age', 'Age_Cat']]
```

```python
column_list = ['Gender', 'Nationality', 'Emirate', 'Cancer_Type', 'Outcome', 'Smoking_Status', 'Comorbidities', 'Ethnicity']

def plot_bar_graph():
    for c in column_list:
        g = sns.catplot(x=c, data=df, kind='count')
        g.set_xticklabels(rotation=90)
        plt.show()

plot_bar_graph()
```

```python
col = ['Age', 'Weight', 'Height']
def plot_bar_graph():
    for c in col:
        sns.distplot(df[c])
        plt.show()
plot_bar_graph()
```
